```python
import pandas as pd
```

```python
df = pd.read_csv('/Users/given/projects/BioRiskEval_pLM/bioriskeval/mut/merged_proteingym_nucleotides_selected_with_source.csv')
```

```python
df.head()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 시퀀스 길이 계산
sequence_lengths = df["mutated_sequence"].apply(len)

# 히스토그램 그리기
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.hist(sequence_lengths, bins=50, edgecolor='black', alpha=0.7)
plt.xlabel('Sequence Length')
plt.ylabel('Count')
plt.title('Distribution of Mutated Sequence Lengths')
plt.axvline(x=1024, color='r', linestyle='--', linewidth=2, label='1024 threshold')
plt.legend()
plt.grid(True, alpha=0.3)

# 1024를 넘는 비율 계산
over_1024 = (sequence_lengths > 1024).sum()
total = len(sequence_lengths)
ratio = over_1024 / total * 100

# 통계 정보
plt.subplot(1, 2, 2)
plt.axis('off')
stats_text = f"""
Sequence Length Statistics:
━━━━━━━━━━━━━━━━━━━━━━━━━━━━
Total sequences: {total:,}
Mean length: {sequence_lengths.mean():.2f}
Median length: {sequence_lengths.median():.2f}
Min length: {sequence_lengths.min()}
Max length: {sequence_lengths.max()}
Std deviation: {sequence_lengths.std():.2f}

━━━━━━━━━━━━━━━━━━━━━━━━━━━━
Sequences > 1024: {over_1024:,}
Percentage > 1024: {ratio:.2f}%
Sequences ≤ 1024: {total - over_1024:,}
Percentage ≤ 1024: {100 - ratio:.2f}%
"""
plt.text(0.1, 0.5, stats_text, fontsize=11, family='monospace', verticalalignment='center')

plt.tight_layout()
plt.show()

# 간단한 출력
print(f"\n{'='*50}")
print(f"Total sequences: {total:,}")
print(f"Sequences with length > 1024: {over_1024:,} ({ratio:.2f}%)")
print(f"Sequences with length ≤ 1024: {total - over_1024:,} ({100 - ratio:.2f}%)")
print(f"{'='*50}")
```

```python
# 1000개 샘플링
sampled_df = df.sample(n=1000, random_state=42)

# 저장 경로
output_path = '/Users/given/projects/BioRiskEval_pLM/bioriskeval/mut/merged_sampled.csv'

# CSV로 저장
sampled_df.to_csv(output_path, index=False)

print(f"샘플링 완료!")
print(f"원본 데이터: {len(df):,}개")
print(f"샘플링된 데이터: {len(sampled_df):,}개")
print(f"저장 위치: {output_path}")
```

```python
# Source 분포 비교 (전체 vs 샘플)
fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# 1. 전체 데이터 - Source별 카운트
source_counts_full = df['source'].value_counts()
axes[0, 0].bar(range(len(source_counts_full)), source_counts_full.values, color='steelblue', alpha=0.7)
axes[0, 0].set_xlabel('Source', fontsize=10)
axes[0, 0].set_ylabel('Count', fontsize=10)
axes[0, 0].set_title(f'Full Dataset - Source Distribution (Total: {len(df):,})', fontsize=12, fontweight='bold')
axes[0, 0].set_xticks(range(len(source_counts_full)))
axes[0, 0].set_xticklabels(source_counts_full.index, rotation=45, ha='right', fontsize=8)
axes[0, 0].grid(True, alpha=0.3, axis='y')

# 2. 전체 데이터 - Source별 비율 (파이 차트)
cmap = plt.get_cmap('Set3')
colors = [cmap(i) for i in range(len(source_counts_full))]
axes[0, 1].pie(source_counts_full.values, labels=source_counts_full.index, autopct='%1.1f%%', 
               colors=colors, textprops={'fontsize': 8})
axes[0, 1].set_title('Full Dataset - Source Percentage', fontsize=12, fontweight='bold')

# 3. 샘플링 데이터 - Source별 카운트
source_counts_sampled = sampled_df['source'].value_counts()
axes[1, 0].bar(range(len(source_counts_sampled)), source_counts_sampled.values, color='coral', alpha=0.7)
axes[1, 0].set_xlabel('Source', fontsize=10)
axes[1, 0].set_ylabel('Count', fontsize=10)
axes[1, 0].set_title(f'Sampled Dataset - Source Distribution (Total: {len(sampled_df):,})', fontsize=12, fontweight='bold')
axes[1, 0].set_xticks(range(len(source_counts_sampled)))
axes[1, 0].set_xticklabels(source_counts_sampled.index, rotation=45, ha='right', fontsize=8)
axes[1, 0].grid(True, alpha=0.3, axis='y')

# 4. 샘플링 데이터 - Source별 비율 (파이 차트)
axes[1, 1].pie(source_counts_sampled.values, labels=source_counts_sampled.index, autopct='%1.1f%%',
               colors=colors, textprops={'fontsize': 8})
axes[1, 1].set_title('Sampled Dataset - Source Percentage', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()

# 상세 통계 출력
print("\n" + "="*80)
print("SOURCE DISTRIBUTION COMPARISON")
print("="*80)
print("\n[FULL DATASET]")
print("-"*80)
for source, count in source_counts_full.items():
    percentage = (count / len(df)) * 100
    print(f"{source:50s} | Count: {count:6,} | Percentage: {percentage:6.2f}%")

print("\n[SAMPLED DATASET]")
print("-"*80)
for source, count in source_counts_sampled.items():
    percentage = (count / len(sampled_df)) * 100
    print(f"{source:50s} | Count: {count:6,} | Percentage: {percentage:6.2f}%")
print("="*80)
```

```python
# 샘플링된 데이터의 시퀀스 길이 계산
sampled_sequence_lengths = sampled_df["mutated_sequence"].apply(len)

# 히스토그램 그리기
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.hist(sampled_sequence_lengths, bins=50, edgecolor='black', alpha=0.7)
plt.xlabel('Sequence Length')
plt.ylabel('Count')
plt.title('Distribution of Sampled Sequence Lengths (n=1000)')
plt.axvline(x=1024, color='r', linestyle='--', linewidth=2, label='1024 threshold')
plt.legend()
plt.grid(True, alpha=0.3)

# 1024를 넘는 비율 계산
sampled_over_1024 = (sampled_sequence_lengths > 1024).sum()
sampled_total = len(sampled_sequence_lengths)
sampled_ratio = sampled_over_1024 / sampled_total * 100

# 통계 정보
plt.subplot(1, 2, 2)
plt.axis('off')
stats_text = f"""
Sequence Length Statistics:
━━━━━━━━━━━━━━━━━━━━━━━━━━━━
Total sequences: {sampled_total:,}
Mean length: {sampled_sequence_lengths.mean():.2f}
Median length: {sampled_sequence_lengths.median():.2f}
Min length: {sampled_sequence_lengths.min()}
Max length: {sampled_sequence_lengths.max()}
Std deviation: {sampled_sequence_lengths.std():.2f}

━━━━━━━━━━━━━━━━━━━━━━━━━━━━
Sequences > 1024: {sampled_over_1024:,}
Percentage > 1024: {sampled_ratio:.2f}%
Sequences ≤ 1024: {sampled_total - sampled_over_1024:,}
Percentage ≤ 1024: {100 - sampled_ratio:.2f}%
"""
plt.text(0.1, 0.5, stats_text, fontsize=11, family='monospace', verticalalignment='center')

plt.tight_layout()
plt.show()

# 간단한 출력
print(f"\n{'='*50}")
print(f"Total sequences: {sampled_total:,}")
print(f"Sequences with length > 1024: {sampled_over_1024:,} ({sampled_ratio:.2f}%)")
print(f"Sequences with length ≤ 1024: {sampled_total - sampled_over_1024:,} ({100 - sampled_ratio:.2f}%)")
print(f"{'='*50}")
```
